## Mô hình Linear Classification

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Training data (binary classification)
X = np.array([0, 1, 2, 3], dtype=float)
Y = np.array([0, 0, 0, 1], dtype=float)  # class labels 0 hoặc 1

# Parameters
w = 0.0
b = 0.0
lr = 0.1
max_iters = 1000

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def loss_fn(X, Y, w, b):
    y_hat = sigmoid(w * X + b)
    return -np.mean(Y * np.log(y_hat + 1e-9) + (1 - Y) * np.log(1 - y_hat + 1e-9))

# Lưu loss theo từng bước
loss_history = []

for step in range(max_iters):

    # forward
    z = w * X + b
    y_hat = sigmoid(z)

    # gradients
    dw = np.mean((y_hat - Y) * X)
    db = np.mean(y_hat - Y)

    # logging
    if step % 50 == 0:
        print("-----------------------------------------------------")
        print("Step %d:" % step)
        print("  w = %.6f, b = %.6f" % (w, b))
        print("  Predictions:", y_hat)
        print("  Loss = %.6f" % loss_fn(X, Y, w, b))
        print("  Gradient dw = %.6f, db = %.6f" % (dw, db))

    # lưu loss
    loss_history.append(loss_fn(X, Y, w, b))

    # update
    w -= lr * dw
    b -= lr * db

# Final parameters
print("\nKết quả cuối:")
print("  w = %.6f, b = %.6f" % (w, b))
print("  Predictions:", sigmoid(w * X + b))
print("  Class:", (sigmoid(w * X + b) >= 0.5).astype(int))

plt.figure(figsize=(8,5))
plt.plot(loss_history, label='Loss')
plt.xlabel('Iteration')
plt.ylabel('Loss')
plt.title('Biểu đồ hội tụ')
plt.grid(True)
plt.legend()
plt.show()
```
