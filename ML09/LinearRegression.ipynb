## Mô hình hồi quy tuyến tính

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Training Data (ví dụ đơn giản)
X = np.array([1, 2, 3], dtype=float)
Y = np.array([2, 4, 6], dtype=float)

# Parameters
w = 0.0
b = 0.0
lr = 0.1
max_iters = 50

N = len(X)
loss_history = []

def mse_loss(X, Y, w, b):
    preds = w * X + b
    return np.mean((Y - preds)**2)

for step in range(max_iters):
    preds = w * X + b              # y_hat
    errors = preds - Y             # (wx + b - y)
    
    # Gradients (derivative)
    dw = (2 / N) * np.sum(errors * X)
    db = (2 / N) * np.sum(errors)

    # In thông tin bước học
    print("------------------------------------------------------")
    print("Step %d:" % step)
    print("  w = %.6f, b = %.6f" % (w, b))
    print("  Predictions:", preds)
    print("  Loss = %.6f" % mse_loss(X, Y, w, b))
    print("  Gradient dw = %.6f, db = %.6f" % (dw, db))

    loss_history.append(mse_loss(X, Y, w, b))

    # Update parameters
    w = w - lr * dw
    b = b - lr * db

    print("  Updated w = %.6f, b = %.6f" % (w, b))

print("\nKết quả cuối:")
print("w = %.6f, b = %.6f" % (w, b))
print("Kết quả dự đoán:", w * X + b)

plt.figure(figsize=(8,5))
plt.plot(loss_history, label='Loss')
plt.xlabel('Iteration')
plt.ylabel('Loss')
plt.title('Biểu đồ hội tụ')
plt.grid(True)
plt.legend()
plt.show()
```
