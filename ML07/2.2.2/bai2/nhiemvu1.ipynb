# Nhiệm vụ 1: Tìm hiểu về cách biểu diễn và hiển thị các ảnh từ tập dữ liệu là hình ảnh

## 1. Import thư viện và tải dữ liệu MNIST

```python
# Import thư viện cần thiết cho việc hiển thị và xử lý dữ liệu
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
%matplotlib inline
from sklearn.datasets import load_digits  # Tải dữ liệu MNIST từ Scikit-learn

# Tải dữ liệu chữ viết tay MNIST (0-9)
digits = load_digits(n_class=10)

# Kiểm tra thông tin tổng quan về dữ liệu
print("Thông tin dữ liệu MNIST:")
print(digits)
```

## 2. Khảo sát thông tin có trong digits

```python
# Xem thông tin của một hình ảnh dưới dạng ma trận 8x8
print("Ma trận 8x8 của hình ảnh đầu tiên:")
print(digits['data'][0].reshape(8, 8))

# Xem thông tin của một hình ảnh dưới dạng mảng 1D (64 pixels)
print("\nMảng 1D của hình ảnh đầu tiên:")
print(digits['data'][0])

# Xem 9 nhãn đầu tiên
print("\n9 nhãn đầu tiên:")
print(digits['target'][0:9])
```

## 3. Vẽ 64 hình ảnh đầu tiên từ ma trận 8x8

```python
# Mỗi chữ số được biểu diễn dưới dạng ma trận 8x8 = 64 pixels
# Mỗi giá trị trong 64 pixel đại diện cho mức độ xám
# Sử dụng imshow để vẽ với thang màu phù hợp

fig, ax = plt.subplots(8, 8, figsize=(10, 10))
for i, axi in enumerate(ax.flat):
    axi.imshow(digits.images[i], cmap='binary')  # cmap='binary' cho ảnh đen trắng
    axi.set(xticks=[], yticks=[])  # Ẩn các tick trên trục

plt.suptitle('64 Hình ảnh chữ viết tay đầu tiên từ MNIST', fontsize=16)
plt.tight_layout()
plt.show()
```

## 4. Vẽ một ảnh cụ thể từ ma trận 8x8

```python
# Hàm để vẽ một ảnh có kích thước 8x8 từ dữ liệu images
def view_digit(index):
    plt.figure(figsize=(4, 4))
    plt.imshow(digits.images[index], cmap=plt.cm.gray_r)  # gray_r cho màu xám đảo ngược
    plt.title('Original it is: ' + str(digits.target[index]))
    plt.axis('off')  # Ẩn trục tọa độ
    plt.show()

# Vẽ ảnh ở vị trí thứ 4 (index 4)
view_digit(4)
```
