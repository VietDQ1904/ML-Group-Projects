# Dự đoán trên test Titanic (Prediction)

## Import thư viện

```python
import pandas as pd
import joblib
```

## 1. Mục tiêu dự đoán
+ Load best model.
+ Predict trên test processed.
+ Tạo submission.csv.

## 2. Load data và model

```python
# Load test processed
test_path = '../data/processed/test_processed.csv'
df_test = pd.read_csv(test_path)

# Load best model
best_model = joblib.load('../models/saved_models/best_svc_model.pkl')

# Predict và ép kiểu sang int
predictions = best_model.predict(df_test).astype(int)  # Thêm .astype(int)

# Load original test để lấy PassengerId
test_orig_path = '../data/raw/test.csv'
df_test_orig = pd.read_csv(test_orig_path)

# Tạo submission
submission = pd.DataFrame({'PassengerId': df_test_orig['PassengerId'], 'Survived': predictions})
submission.to_csv('../submission.csv', index=False)
print("Submission created.")
```
