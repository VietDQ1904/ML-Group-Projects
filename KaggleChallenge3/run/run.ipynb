## 1. Định nghĩa vấn đề
+ **Mục tiêu**:
   - Sử dụng mô hình đã huấn luyện và lưu ở phần 4 để dự đoán thể loại nhạc trên tập test của Kaggle.
   - Xuất ra file submission và đăng lên Kaggle.



## 2 Chuẩn bị vấn đề

### 2.1. Import các thư viện cần thiết

```python
import numpy as np
import pickle
import pandas as pd
import joblib
```

### 2.2. Lấy tập test đã xử lý

```python
test_data = pd.DataFrame(pd.read_pickle("../preprocessing/data/test_processed.pkl"))
```

```python
test_data.head()
```

## 3. Tải mô hình tốt nhất đã được huấn luyện

```python
best_model = joblib.load("../models/saved_model_stacking.pkl") 
```

## 4. Thực hiện dự đoán trên tập test

```python
y_test_pred = best_model.predict(test_data)
```

## 5. Lưu ra file submission theo Kaggle

```python
# Đọc file csv từ file sample
sample = pd.read_csv("../data/shai-music-genre-classification/sample_submission (2).csv")

# Lưu file submission đúng dạng Kaggle
submission = pd.DataFrame({
    "Id": sample["Id"],
    "Class": y_test_pred
})

submission.to_csv("submission.csv", index=False) # Xuất file submission
print("Đã tạo ra file submission.csv")
```
