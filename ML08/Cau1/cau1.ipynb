```python
import pandas as pd
import numpy as np
from collections import Counter

# ----------------------------
# 1. Load dữ liệu từ file CSV
# ----------------------------
file_path = r"C:\Users\Admin\OneDrive\Desktop\ML Nhóm\ML-Group-Projects\ML08\Cau1\data\iris\iris.data"
df = pd.read_csv(file_path)

# Giả sử cột cuối là label (species), các cột còn lại là đặc trưng
X = df.iloc[:, :-1].values   # features
y = df.iloc[:, -1].values    # labels


# ----------------------------
# 2. Hàm chia dữ liệu train / test
# ----------------------------
def train_test_split_custom(X, y, test_size=0.3):
    np.random.seed(42)
    indices = np.random.permutation(len(X))
    test_cnt = int(len(X) * test_size)

    test_idx = indices[:test_cnt]
    train_idx = indices[test_cnt:]

    return X[train_idx], X[test_idx], y[train_idx], y[test_idx]


X_train, X_test, y_train, y_test = train_test_split_custom(X, y)


# ----------------------------
# 3. Hàm tính khoảng cách Euclid
# ----------------------------
def euclidean_distance(x1, x2):
    return np.sqrt(np.sum((x1 - x2) ** 2))


# ----------------------------
# 4. Tự cài thuật toán KNN
# ----------------------------
def knn_predict(X_train, y_train, x_test, k):
    distances = []

    for i in range(len(X_train)):
        dist = euclidean_distance(X_train[i], x_test)
        distances.append((dist, y_train[i]))

    # sắp xếp theo khoảng cách tăng dần rồi lấy k hàng xóm gần nhất
    neighbors = sorted(distances, key=lambda x: x[0])[:k]

    # lấy danh sách nhãn trong k hàng xóm
    labels = [label for _, label in neighbors]

    # bỏ phiếu majority vote
    prediction = Counter(labels).most_common(1)[0][0]
    return prediction


# ----------------------------
# 5. Kiểm tra mô hình trên tập test
# ----------------------------
def accuracy_score_custom(y_true, y_pred):
    return np.sum(y_true == y_pred) / len(y_true)


k = 5
predictions = []

for sample in X_test:
    pred = knn_predict(X_train, y_train, sample, k)
    predictions.append(pred)

acc = accuracy_score_custom(np.array(y_test), np.array(predictions))


print("=== KẾT QUẢ KNN (TỰ CÀI) ===")
print("K =", k)
print("Accuracy =", acc)
```
